# 01 · Exploratory data analysis
Run `python scripts/download_data.py` first. Everything below reads the saved outputs.

```
import json, pandas as pd
from pathlib import Path
ROOT = Path('..').resolve()
insp = json.loads((ROOT/'reports/tables/dataset_inspection.json').read_text())
for name, d in insp['datasets'].items():
    print(name, d.get('raw', {}).get('rows'), d.get('raw', {}).get('class_distribution'), d.get('cleaning'))
```

```
train, val, test = (pd.read_parquet(ROOT/f'data/processed/{s}.parquet') for s in ('train','val','test'))
print(len(train), len(val), len(test))
for n, d in (('train',train),('val',val),('test',test)):
    print(n, d['label'].value_counts().to_dict())
```
